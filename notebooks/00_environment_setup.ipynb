Lấy source code từ GitHub

```python
from google.colab import userdata
import subprocess
import os

token = userdata.get("GITHUB_TOKEN")

repo_url = "https://github.com/PhuocManh01/HGNN-Attention-Recommender.git"

env = os.environ.copy()
env["GIT_CONFIG_COUNT"] = "1"
env["GIT_CONFIG_KEY_0"] = "http.extraHeader"

import base64

credentials = base64.b64encode(
    f"x-access-token:{token}".encode()
).decode()

env["GIT_CONFIG_VALUE_0"] = (
    f"AUTHORIZATION: basic {credentials}"
)

subprocess.run(
    ["git", "clone", repo_url],
    env=env,
    check=True
)
```

Kiểm tra PyTorch Geometric

```python
import torch
import torch_geometric

from torch_geometric.data import HeteroData

graph = HeteroData()

graph["user"].num_nodes = 2
graph["movie"].num_nodes = 3

graph["user", "rates", "movie"].edge_index = (
    torch.tensor([
        [0, 1],
        [1, 2]
    ], dtype=torch.long)
)

print("PyG:", torch_geometric.__version__)
print(graph)
print("Heterogeneous Graph: PASSED")
```
